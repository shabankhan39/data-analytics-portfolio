```python
from bs4 import BeautifulSoup
import requests
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'


headers = {
    'User-Agent': 'DataScrapingPracticeBot/1.0 (your-email@example.com)'
}
page = requests.get(url, headers=headers)
print("Response Status:", page)
```

```python
soup=BeautifulSoup(page.text,'html')
table1=soup.find_all('table')[0]
```

```python
title=table1.find_all('th')
```

```python
title
```

```python
world_title= [tit.text.strip() for tit in title]
```

```python
world_title
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(columns=world_title)
```

```python
df
```

```python
columns_data=table1.find_all('tr')
```

```python
for rows in columns_data[1:]:
    row_data = rows.find_all('td')
    individual_row_data= [data.text.split() for data in row_data]
    length=len(df)
    df.loc[length]=individual_row_data
```
