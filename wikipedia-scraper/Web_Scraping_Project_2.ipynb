```python
from bs4 import BeautifulSoup
import requests
```

```python
url='https://en.wikipedia.org/wiki/The_World%27s_Billionaires'
```

```python
headers = {
    'User-Agent': 'DataScrapingPracticeBot/1.0 (your-email@example.com)'
}
page = requests.get(url, headers=headers)
```

```python
soup = BeautifulSoup(page.text,'html')
```

```python
table=soup.find_all('table' , class_="wikitable sortable" , id="mwkA")[0]
```

```python
t2026=table.find_all('th')
```

```python
t2026_headers=[head.text.strip() for head in t2026]
```

```python
t2026_headers
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(columns=t2026_headers)
```

```python
table_rows=table.find_all('tr')
```

```python
(df)
```

```python
for rows in table_rows[1:11]:
    rows_data= rows.find_all('td')
    individual_rows_data= [data.text.strip() for data in rows_data]
    print(individual_rows_data)
    length = len(df)
    df.loc[length] = individual_rows_data
```
